```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel("HR Analyst.xlsx")
```

```python
print(df.head())
```

```python
age_attrition = pd.crosstab(
    df["CF_age band"],
    df["Attrition"]
)

print(age_attrition)
```

```python
age_attrition.plot(kind="bar")
plt.title("CF_age band vs attrition")
plt.xlabel("CF_age band")
plt.ylabel("Number of Employees")
plt.xticks(rotation=0)
plt.show()
```

```python
salary_attrition = df.groupby("Attrition")["Monthly Income"].mean()

print(salary_attrition)
```

```python
salary_attrition.plot(kind="bar")

plt.title("Average Monthly Income by Attrition")
plt.xlabel("Attrition")
plt.ylabel("Average Monthly Income")
plt.xticks(rotation=0)

plt.show()
```

```python
overtime_attrition = pd.crosstab(
    df["Over Time"],
    df["Attrition"]
)

print(overtime_attrition)
```

```python
overtime_attrition.plot(kind="line", marker="o")

plt.title("Overtime vs Attrition")
plt.xlabel("Overtime")
plt.ylabel("Number of Employees")
plt.legend(title="Attrition")
plt.show()
```
